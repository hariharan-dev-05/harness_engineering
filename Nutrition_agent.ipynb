### Groq Client Setup
Please make sure to add your Groq API key to the Colab Secrets manager (the key icon 🔑 on the left panel) with the name `GROQ_API_KEY` before running the next cell.

```python
from groq import Groq
from google.colab import userdata

# Securely retrieve the Groq API key from Colab Secrets
try:
    groq_api_key = userdata.get('GROQ_API_KEY')
    client = Groq(api_key=groq_api_key)
    print("Groq client successfully initialized!")
except Exception as e:
    print(f"Error initializing Groq client: {e}\nPlease ensure GROQ_API_KEY is configured in your Secrets.")
```

### Nutrition Agent Helper Function
We will define a helper function `analyze_meal` that takes a description of food, passes it to the Llama-3 model on Groq, and requests a detailed nutritional breakdown.

```python
def analyze_meal(meal_description):
    system_prompt = """
    You are a professional Nutritionist Agent. Your task is to analyze the user's meal description and provide:
    1. An estimated breakdown of total calories and macronutrients (protein, carbohydrates, fats).
    2. A brief analysis of nutritional benefits and health recommendations.
    3. Suggestions on how to optimize this meal (e.g., adding more fiber, reducing sodium).
    Be precise, scientific, yet encouraging and easy to understand.
    """

    try:
        completion = client.chat.completions.create(
            model="openai/gpt-oss-120b",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": meal_description}
            ],
            temperature=0.2
        )
        return completion.choices[0].message.content
    except Exception as e:
        return f"An error occurred during API request: {e}"
```

```python
# Example usage:
meal = "A bowl of oatmeal with 1 banana, a handful of almonds, and a scoop of whey protein powder."
analysis = analyze_meal(meal)
print(analysis)
```
